```python
Consumer_Name = input("Consumer Name :")
Consumer_Number = int(input("Consumer Number :"))
Tariff = input("Tariff (Domestic / Commercial) :")
Tariff = Tariff.strip().lower()
Units = int(input("Consumed Units :"))
Bill_Payment = int(input("Bill Payment :"))
CC_Subsidy = int(input("CC Subsidy Amount :"))
New_Subsidy = int(input("New Subsidy Amount :"))
```

```python
if Tariff == "domestic":
    if Units <= 200:
        Total = 0
    else:
        Total = (Units - 200) * 8

elif Tariff == "commercial":
    Total = Units * 10

else:
    print("Invalid Tariff! Enter Domestic or Commercial.")
    Total = None

if Total is not None:
    Net_Amount = Total - CC_Subsidy - New_Subsidy
    
    print("---------------------------------------")
    print("TNEB ELECTRICITY BILL")
    print("---------------------------------------")
    print("Consumer Name:", Consumer_Name)
    print("Consumer Number:", Consumer_Number)
    print("Tariff:", Tariff.title())
    print("Consumed Units:", Units)
    print("EB Bill (Rs.):", Total)
    print("Bill Payment (Rs.):", Bill_Payment)
    print("CC Subsidy (Rs.):", CC_Subsidy)
    print("New Subsidy (Rs.):", New_Subsidy)
    print("Total Current Charges (Rs.):", Net_Amount)
```
